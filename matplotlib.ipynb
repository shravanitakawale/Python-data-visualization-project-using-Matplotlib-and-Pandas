````python
arr=np.array([1,2,3,4,5])
arr=arr*2
````

````python
arr2 = arr ** 2
plt.plot(arr, arr2, marker='o')
plt.show()
````

````python
plt.plot(arr, arr2, color='red',marker='o',markerfacecolor='orange',markeredgecolor='orange')
plt.grid()
plt.show()
````

````python
plt.plot(arr, arr2, color='red',marker='o',markerfacecolor='orange',markeredgecolor='orange')
plt.grid()
plt.xlabel('array')
plt.ylabel('twice of array')
plt.title('example of line plot')
for i,j in zip(arr,arr2):
  plt.annotate(f'({i},{j})', xy=(i, j))
plt.show()
````

````python
df=pd.read_csv('https://raw.githubusercontent.com/Taj1920/Data-Analysis-Datasets/refs/heads/main/Datasets-Practice/flipkart_sales.csv')
df
````

````python
cat = df.groupby('Category')['Total Sales (INR)'].sum().reset_index()

plt.plot(cat['Category'],cat['Total Sales (INR)'],color='red',marker='o',markerfacecolor='orange',markeredgecolor='orange')
plt.grid()
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.title('Category wise total sales')
for i,j in zip(cat['Category'],cat['Total Sales (INR)']):
  plt.annotate(f'({i},{j})', xy=(i, j))                        #annotate()- display the text on the graph (electronic,173071)
plt.show()
````

```
1.df → your original DataFrame/table.
.groupby('Category') → put rows having the same category together.
['Total Sales (INR)'] → select the Total Sales column.
.sum() → add the sales of each category.
.reset_index() → convert the result back into a normal DataFrame/table.
cat = → save the result in a variable called cat.
cat means category-wise total sales table.

2.plt.plot() → used to create a line graph.

cat['Category']
→ What should be shown on the X-axis.

cat['Total Sales (INR)']
→ What should be shown on the Y-axis.

color='red'
→ Makes the line red.

marker='o'
→ Puts a circle o at every data point.

markerfacecolor='orange'
→ Makes the inside of the circle orange.

markeredgecolor='orange'
→ Makes the border of the circle orange.

3.plt.grid() → adds horizontal and vertical grid lines to the graph.

4.plt.xlabel('Category')
xlabel() → gives a name to the X-axis.
'Category' → the name displayed.

5.plt.ylabel('Total Sales')
ylabel() → gives a name to the Y-axis.
'Total Sales' → the name displayed.

6.Give title to graph
plt.title('Category wise total sales')
title() → adds a title at the top of the graph.

7.zip(cat['Category'], cat['Total Sales (INR)'])
zip() → pairs two columns together.
zip() makes pairs:
[example
(Electronics, 50000)
(Clothing, 30000)
(Food, 20000)]
i → Category
j → Total Sales
then
i = Clothing
j = 30000
so on

8.annotate()
plt.annotate(f'({i},{j})', xy=(i, j))
plt.annotate() → writes text near a particular point on the graph.
f means formatted string.

9.Display the graph
plt.show()
plt.show() → actually displays the graph.

Easy flow to remember:
Data → Group → Sum → Plot → Grid → Labels → Title → Annotate → Show.

```




````python
df['month'] = pd.to_datetime(df['Order Date']).dt.strftime('%B')
cat = df.groupby('month')['Total Sales (INR)'].sum().reset_index()

plt.plot(cat['month'],cat['Total Sales (INR)'],color='red',marker='o',markerfacecolor='orange',markeredgecolor='orange')
plt.grid()
plt.xlabel('month')
plt.xticks(rotation=90)
plt.ylabel('Total Sales')
plt.title('month wise total sales')
for i,j in zip(cat['month'],cat['Total Sales (INR)']):
  plt.annotate(f'({i},{j})', xy=(i, j))
plt.show()
````

````python
#piechart: to represent the percentage of individual data, categorical data
````

````python
subject=['python','ML','Data Analysis','Data Engineering']
#marks=[10,30,40,50]
marks=[35,25,25,25]
plt.pie(marks,labels=subject,colors=['yellow','blue','#F54927','orange'],autopct='%0.2f%%',explode=[0.1,0,0,0])  #html color code
plt.show()
````

1.Create the subject list
subject=['python','ML','Data Analysis','Data Engineering']

This creates a Python list containing the names of 4 subjects.
Python
ML
Data Analysis
Data Engineering
These names will be displayed as labels in the pie chart

2.Create the marks list
marks=[35,25,25,25]
This contains the marks/values corresponding to each subject.
35 + 25 + 25 + 25 = 110
The pie chart converts these values into percentages.

3.Create the pie chart
plt.pie(
    marks,
    labels=subject,
    colors=['yellow','blue','#F54927','orange'],
    autopct='%0.2f%%',
    explode=[0.1,0,0,0]
)

marks
plt.pie(marks)  marks->[35,25,25,25]
Tells Matplotlib which values should be represented in the pie chart.

labels=subject
Adds the subject names to each section.
subject->Python,ML,Data Analysis,Data Engineering

colors=['yellow','blue','#F54927','orange']
Assigns a different color to each section.

4.autopct='%0.2f%%'
This displays the percentage inside each slice.
% → formatting symbol
0.2f → display the number with 2 digits after the decimal
%% → display the % symbol

5.explode=[0.1,0,0,0]
This is used to separate/pull out a slice from the pie chart.
Python           → 0.1
ML               → 0
Data Analysis    → 0
Data Engineering → 0

So Python's slice will move slightly away from the center.
then ML would be separated instead.

6.plt.show()

This tells Python to display the chart.

````python
category_sales = df.groupby('Category')['Total Sales (INR)'].sum().reset_index()

plt.pie(
    category_sales['Total Sales (INR)'],
    labels=category_sales['Category'],
    colors=['yellow', 'blue', '#F54927', 'orange', 'pink'],
    autopct='%0.2f%%',
    explode=[0, 0, 0, 0.1, 0]
)
plt.show()
````
